## Title

# Introduction

Machine Learning is a branch of Artificial Intelligence that enables computers to learn from data and make predictions. Classification is one of the important tasks in machine learning.

In this project, the Iris flower dataset is used to classify flower species using machine learning algorithms.

The project is implemented using Python and scikit-learn.

---

# Objective

The main objective of this project is:

- To understand machine learning classification
- To train machine learning models
- To classify iris flower species
- To evaluate model accuracy

# Technologies Used

- Python
- Pandas
- NumPy
- Matplotlib
- scikit-learn
- Google Colab

# Dataset

The Iris dataset is a popular machine learning dataset.

It contains:
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

Target Classes:
- Iris Setosa
- Iris Versicolor
- Iris Virginica

---

# Algorithms Used

## 1. Logistic Regression
Logistic Regression is a supervised machine learning algorithm used for classification tasks.

## 2. Random Forest Classifier
Random Forest is an ensemble learning algorithm that combines multiple decision trees for better accuracy.
